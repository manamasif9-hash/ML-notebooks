## 1️⃣ Pandas Series — The Building Block

A **Series** is just a single column of data, like one column in Excel.
It has:
- **values** (the actual data)
- an **index** (labels for each value, like row numbers)

Think of it as a smart list.

```python
# Creating a Series from a normal Python list
# pandas automatically creates an index 0,1,2,3... for us
marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
print()
print("Data type of marks:", type(marks))
```

```python
# We can give our own custom index (labels) instead of 0,1,2,3...
# Here we use student names as the index
marks_named = pd.Series([85, 90, 78, 92, 88],
                         index=["Ali", "Sara", "Bilal", "Ayesha", "Zain"])

print(marks_named)
print()

# Accessing a single value using its label (just like a dictionary!)
print("Sara's marks:", marks_named["Sara"])
```

## 2️⃣ DataFrame — The Real Power of Pandas

A **DataFrame** is a full table — many rows and many columns, just like an Excel sheet
or a SQL table. This is what you will use 90% of the time in real projects.

We can build a DataFrame from a Python **dictionary**, where:
- each **key** becomes a column name
- each **value** (a list) becomes the data in that column

```python
# Creating a DataFrame using a dictionary
data = {
    "Name":   ["Ali", "Sara", "Bilal", "Ayesha", "Zain"],
    "Age":    [20, 22, 21, 23, 20],
    "City":   ["Lahore", "Karachi", "Faisalabad", "Islamabad", "Multan"],
    "Marks":  [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

# Always print/display your DataFrame to check it looks correct
df
```

```python
# .head() shows the first 5 rows (very useful for big datasets)
# .tail() shows the last 5 rows
print("First 3 rows:")
print(df.head(3))

print("\nLast 2 rows:")
print(df.tail(2))
```

```python
# .info() gives a summary: column names, data types, non-null counts
# This is usually the FIRST thing a real data analyst runs on new data
df.info()
```

```python
# .describe() gives statistical summary of NUMERIC columns only
# count, mean, std (standard deviation), min, max, and quartiles
df.describe()
```

```python
# .shape tells us (rows, columns) -> very quick way to know dataset size
print("Shape of dataframe (rows, columns):", df.shape)

# .columns tells us the column names
print("Column names:", df.columns.tolist())

# .dtypes tells us the data type of each column
print("\nData types:\n", df.dtypes)
```

## 3️⃣ Selecting Rows & Columns

This is the MOST used skill in pandas. There are 2 main ways:
- `df['column']` → select one column
- `df.loc[]` → select by **label**
- `df.iloc[]` → select by **position number** (like array indexing)

```python
# Selecting a SINGLE column -> returns a Series
print(df["Name"])
print(type(df["Name"]))  # Notice: single column = Series
```

```python
# Selecting MULTIPLE columns -> pass a LIST of column names -> returns a DataFrame
print(df[["Name" ,"Marks"]])
print(type(df[["Name","Marks"]])) # Notice: multiple columns = DataFrame
```

```python
# .loc -> select rows/columns by LABEL (index name / column name)
# Syntax: df.loc[row_label, column_label]

# Get row with index 2 (Bilal's row)
print(df.loc[2])

print()

# Get specific rows and specific columns
print(df.loc[0:2, ["Name", "Score"]])   # rows 0 to 2 (inclusive!), only Name & Marks columns
```

```python
# .iloc -> select rows/columns by POSITION (integer index, like normal lists)
# Syntax: df.iloc[row_position, column_position]

print(df.iloc[0])          # first row
print()
print(df.iloc[0:3, 0:2])   # first 3 rows, first 2 columns (end is EXCLUSIVE here, unlike .loc)
```

### ⚠️ Teacher's Tip (Important Gotcha!)
- `.loc` → label-based, **end index is INCLUDED**
- `.iloc` → position-based, **end index is EXCLUDED** (like normal Python slicing)

This confuses every beginner at least once — now you know it in advance! 😉

## 4️⃣ Filtering Data (Boolean Indexing)

This is how we ask pandas: *"Give me only the rows where a condition is TRUE"*
This is one of the most powerful and commonly used features in pandas.

```python
# Filter: students who scored more than 85
high_scorers = df[df["Marks"] > 85]
print(high_scorers)
```

```python
# Combine multiple conditions
# Use & for AND, | for OR — and ALWAYS wrap each condition in ( )

# Students older than 20 AND marks above 80
result = df[(df["Age"] > 20) & (df["Marks"] > 80)]
print(result)
```

```python
# .isin() -> check if value is inside a list of options
cities_wanted = df[df["City"].isin(["Lahore", "Karachi"])]
print(cities_wanted)
```

## 5️⃣ Adding, Modifying & Dropping Columns

Real datasets are rarely perfect — you'll constantly need to create new columns,
update existing ones, or remove ones you don't need.

```python
# Adding a NEW column: Grade based on Marks
# We use a simple condition here; later we'll use apply() for more complex logic
df["Passed"] = df["Marks"] >= 80   # This creates a column of True/False
print(df)
```

```python
# apply() lets us run a custom function on every value in a column
def get_grade(marks):
    # A simple grading function
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    else:
        return "C"

# axis is not needed here since we're applying to a single column (Series)
df["Grade"] = df["Marks"].apply(get_grade)
print(df)
```

```python
# Dropping a column -> use drop(), specify axis=1 (columns) or columns=[...]
# NOTE: by default this does NOT change the original df unless inplace=True
df_dropped = df.drop(columns=["Passed"])
print(df_dropped)

print()
print("Original df still has 'Passed' column:")
print(df.columns.tolist())
```

```python
# Renaming columns
df_renamed = df.rename(columns={"Marks": "Score", "Name": "Student_Name"})
print(df_renamed.columns.tolist())
```

## 6️⃣ Sorting and Grouping (GroupBy = Superpower)

`groupby()` is one of the most powerful pandas tools — it lets us group rows
by a category and calculate summary statistics for each group (like Excel Pivot Tables).

```python
# Sorting by a single column (ascending by default)
sorted_df = df.sort_values("Marks", ascending=False)  # highest marks first
print(sorted_df)
```

```python
# Sorting by MULTIPLE columns
sorted_multi = df.sort_values(["Grade", "Marks"], ascending=[True, False])
print(sorted_multi)
```

```python
# groupby() example: average marks PER grade
grouped = df.groupby("Grade")["Marks"].mean()
print(grouped)

print()

# groupby with multiple statistics at once using .agg()
grouped_stats = df.groupby("Grade")["Marks"].agg(["mean", "min", "max", "count"])
print(grouped_stats)
```

## 7️⃣ Handling Missing Data

Real-world data is messy — it almost always has missing values (`NaN`).
Knowing how to detect and handle them is a critical skill.

```python
# Let's create a messy dataset with some missing values
messy_data = {
    "Name": ["Ali", "Sara", "Bilal", "Ayesha", np.nan],
    "Marks": [85, np.nan, 78, 92, 88],
    "City": ["Lahore", "Karachi", np.nan, "Islamabad", "Multan"]
}
messy_df = pd.DataFrame(messy_data)
print(messy_df)
```

```python
# Detect missing values -> returns True/False for each cell
print(messy_df.isnull())

print()

# Count missing values PER COLUMN (very commonly used!)
print("Missing values per column:")
print(messy_df.isnull().sum())
```

```python
# Option 1: Drop rows that have ANY missing value
cleaned_drop = messy_df.dropna()
print("After dropna():")
print(cleaned_drop)
```

```python
# Option 2: Fill missing values instead of dropping them
# Fill missing Marks with the average of the column (common technique!)
filled_df = messy_df.copy()
filled_df["Marks"] = filled_df["Marks"].fillna(filled_df["Marks"].mean())
filled_df["City"] = filled_df["City"].fillna("Unknown")

print("After fillna():")
print(filled_df)
```

Using everything you learned above, complete this mini real-world task:

**Scenario:** You are given exam results for a class. Do the following:

1. Create a DataFrame with columns: `Name`, `Math`, `Science`, `English` (at least 6 students, your own data)
2. Add a new column `Total` = sum of the 3 subjects
3. Add a new column `Average` = Total / 3
4. Add a new column `Result` = "Pass" if Average >= 40 else "Fail" (use `.apply()`)
5. Sort students by `Total` in descending order
6. Find the topper (student with the highest Total) using `.loc` and `.idxmax()`

*(This combines Series, DataFrame creation, new columns, apply, sorting, and file I/O — everything from this notebook!)*

```python
# 👉 Student: write your Final Challenge solution here
section = {
    "Name" : ["sadia" ,"haleema","khadija","ayesha","hafsa","salma"],
    "Maths" : [89,75,78,88,90,96],
    "Science" :[67,87,64,75,76,97],
    "English" :[93,85,55,56,78,98]
}

df= pd.DataFrame(section)

df

df['Total']=df['Maths'] + df['Science'] + df['English']

df

df['Average'] = df['Total'] / 3
df 

def get_result(average):
    if average >= 40 :
        return "Pass"
    else :
        return "fail"

df["Result"] = df['Average'].apply(get_result)

print(df)

sortttt = df.sort_values(['Total'] , ascending=False)
print(sortttt)

higher = df.iloc[df["Average"].idxmax()]
print(higher)

df.to_csv ("final_results.csv", index=False)
print("File saved successfully")
```
