## 0️⃣ Why do we even visualize data?

**Purpose:** Numbers in a table are hard for the human brain to interpret quickly.
A chart lets you see **patterns, trends, outliers, and comparisons** instantly —
things that would take minutes to spot by scanning raw numbers.

**Matplotlib** is the original and most widely used Python plotting library.
Almost every other Python visualization library (Seaborn, Pandas' `.plot()`)
is actually built ON TOP of Matplotlib.

**Golden rule:** Every chart should have ONE clear purpose — don't just plot
data because you can; plot it because it answers a question.

## 1️⃣ Your First Plot — The Line Chart

**Purpose:** Line charts show how a value changes over a continuous range
(most commonly TIME). This is the most fundamental chart type — start here.

```python
# Simple data: x = days, y = temperature readings
days = [1, 2, 3, 4, 5, 6, 7]
temperature = [22, 24, 19, 23, 25, 21, 20]

# plt.plot(x, y) draws a line connecting the points
plt.plot(days, temperature)

# ALWAYS label your axes — a chart without labels is meaningless to anyone else!
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature")

# plt.show() actually displays the chart (in scripts; in notebooks it often auto-displays,
# but it's good practice to always call it explicitly)
plt.show()
```

```python
# Let's make it more informative with styling
plt.plot(days, temperature,
         color="orange",      # line color
         linestyle="--",      # dashed line
         marker="o",          # circle marker at each data point
         linewidth=2,         # thickness of the line
         label="City A")      # label used in the legend

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature - Styled")
plt.legend()          # PURPOSE: shows what each line represents (needs label= above)
plt.grid(True)         # PURPOSE: gridlines make it easier to read exact values
plt.show()
```

## 2️⃣ Multiple Lines on One Chart

**Purpose:** Comparing two or more trends side-by-side is one of the most common
real-world visualization tasks — e.g. comparing sales of 2 products over time.

```python
days = np.arange(1, 8)
city_a = [22, 24, 19, 23, 25, 21, 20]
city_b = [18, 20, 17, 19, 22, 18, 16]

# Simply call plt.plot() multiple times BEFORE plt.show() to overlay lines
plt.plot(days, city_a, marker="o", label="City A")
plt.plot(days, city_b, marker="s", label="City B")

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Comparison: City A vs City B")
plt.legend()   # essential when you have multiple lines, so viewers know which is which
plt.grid(True, alpha=0.3)   # alpha = transparency, makes gridlines subtle, not distracting
plt.show()
```

## 3️⃣ Bar Charts — Comparing Categories

**Purpose:** Use bar charts when comparing DISCRETE categories (not continuous data)
— e.g. sales per product, population per country, votes per candidate.
Bars make magnitude differences very easy to see at a glance.

```python
subjects = ["Math", "Science", "English", "History", "Art"]
scores = [85, 92, 78, 88, 95]

plt.bar(subjects, scores, color="skyblue", edgecolor="black")

plt.xlabel("Subject")
plt.ylabel("Score")
plt.title("Student Scores by Subject")
plt.show()
```

```python
# Horizontal bar chart -> PURPOSE: better when category names are long (easier to read)
plt.barh(subjects, scores, color="salmon")
plt.xlabel("Score")
plt.title("Student Scores (Horizontal)")
plt.show()
```

```python
# Grouped bar chart -> PURPOSE: compare 2+ categories side by side for each group
students = ["Ali", "Sara", "Bilal"]
math_scores = [85, 90, 78]
science_scores = [88, 85, 82]

x = np.arange(len(students))   # numeric positions for each student: [0, 1, 2]
width = 0.35                   # width of each bar

# Shift bars left/right so they sit side-by-side instead of overlapping
plt.bar(x - width/2, math_scores, width, label="Math", color="steelblue")
plt.bar(x + width/2, science_scores, width, label="Science", color="darkorange")

plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Math vs Science Scores")
plt.xticks(x, students)   # replace numeric x-axis with actual student names
plt.legend()
plt.show()
```

## 4️⃣ Histograms — Understanding Distributions

**Purpose:** A histogram shows how a SINGLE numeric variable is DISTRIBUTED —
i.e., how many values fall into different ranges ("bins"). This is essential
for understanding your data BEFORE doing any analysis (are values normal? skewed? clustered?).

```python
# Generate 1000 random exam scores following a normal (bell-curve) distribution
np.random.seed(42)   # reproducibility
exam_scores = np.random.normal(loc=70, scale=10, size=1000)  # mean=70, std=10

# bins = how many buckets to divide the range into
# PURPOSE: too few bins hides detail, too many bins makes it noisy — 20-30 is often a good start
plt.hist(exam_scores, bins=25, color="mediumpurple", edgecolor="black")

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.show()
```

### ⚠️ Teacher's Tip
A histogram is NOT a bar chart, even though it looks similar!
- Bar chart = comparing separate CATEGORIES (e.g. subjects)
- Histogram = showing the SPREAD of ONE continuous numeric variable (bins are ranges of numbers)

## 5️⃣ Scatter Plots — Relationships Between Two Variables

**Purpose:** Scatter plots show the RELATIONSHIP (correlation) between two numeric
variables — e.g. "does study time affect exam score?" Each point = one observation.

```python
np.random.seed(1)
study_hours = np.random.uniform(1, 10, 50)          # 50 students, random study hours
# Exam score roughly increases with study hours, plus some random noise
exam_score = study_hours * 7 + np.random.normal(0, 8, 50)

plt.scatter(study_hours, exam_score, color="teal", alpha=0.7)

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

```python
# We can encode EXTRA information using color and size of the points!
# PURPOSE: visualize a 3rd and even 4th variable on a 2D chart
np.random.seed(1)
n = 50
x = np.random.rand(n) * 100
y = np.random.rand(n) * 100
colors = np.random.rand(n)        # a 3rd variable mapped to color
sizes = np.random.rand(n) * 500   # a 4th variable mapped to point size

plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap="viridis")
plt.colorbar(label="Color Scale (3rd variable)")   # shows what the colors mean

plt.xlabel("X value")
plt.ylabel("Y value")
plt.title("Scatter Plot with Color & Size Encoding")
plt.show()
```

## 6️⃣ Pie Charts — Showing Proportions

**Purpose:** Pie charts show what PERCENTAGE each category makes up of a WHOLE.

**Teacher's warning:** Pie charts are heavily overused in the real world and are
often HARDER to read accurately than a simple bar chart — human eyes are bad at
comparing angles and areas. Use pie charts ONLY when you have few categories (3-6)
and truly want to emphasize parts of a whole.

```python
categories = ["Rent", "Food", "Transport", "Entertainment", "Savings"]
budget = [1200, 400, 200, 150, 300]

# autopct shows the percentage on each slice, formatted to 1 decimal place
# startangle rotates the whole pie for a cleaner look
plt.pie(budget, labels=categories, autopct="%1.1f%%", startangle=90,
        colors=["#ff9999","#66b3ff","#99ff99","#ffcc99","#c2c2f0"])

plt.title("Monthly Budget Breakdown")
plt.axis("equal")   # PURPOSE: ensures the pie is drawn as a perfect circle, not an oval
plt.show()
```

## 7️⃣ Subplots — Multiple Charts in One Figure

**Purpose:** Sometimes you need to show several related charts SIDE BY SIDE
for easy comparison — e.g. a dashboard showing sales, revenue, and profit together.

```python
# plt.subplots(rows, cols) creates a GRID of charts and returns:
# fig = the overall figure (the container)
# axes = an array of individual chart "slots" to draw into
fig, axes = plt.subplots(1, 2, figsize=(12, 4))   # 1 row, 2 columns, figsize=(width, height) in inches

# Draw on the FIRST subplot (axes[0])
axes[0].plot(days, city_a, marker="o", color="blue")
axes[0].set_title("City A Temperature")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Temp (°C)")

# Draw on the SECOND subplot (axes[1])
axes[1].bar(subjects, scores, color="green")
axes[1].set_title("Subject Scores")
axes[1].set_xlabel("Subject")
axes[1].set_ylabel("Score")

plt.tight_layout()   # PURPOSE: automatically fixes spacing so titles/labels don't overlap
plt.show()
```

```python
# A 2x2 grid example -> PURPOSE: build a small "dashboard" of 4 different charts
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(days, city_a, color="blue")
axes[0, 0].set_title("Line Chart")

axes[0, 1].bar(subjects, scores, color="orange")
axes[0, 1].set_title("Bar Chart")

axes[1, 0].hist(exam_scores, bins=20, color="purple")
axes[1, 0].set_title("Histogram")

axes[1, 1].scatter(study_hours, exam_score, color="teal")
axes[1, 1].set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```

## 9️⃣ Styling & Customization

**Purpose:** A well-styled chart looks professional and communicates faster.
Small touches like text annotations, custom limits, and themes make a big difference.

```python
# Using a built-in style theme -> PURPOSE: instantly makes charts look more polished/professional
plt.style.use("seaborn-v0_8-darkgrid")   # try changing this to 'ggplot', 'dark_background', 'bmh', etc.

plt.plot(days, city_a, marker="o", label="City A", linewidth=2)
plt.plot(days, city_b, marker="s", label="City B", linewidth=2)

plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.title("Styled Chart Example")
plt.legend()

# Setting explicit axis limits -> PURPOSE: controls exactly what range is visible
plt.ylim(10, 30)

# Annotating a specific point -> PURPOSE: draw attention to something important (e.g. a peak)
plt.annotate("Peak!", xy=(5, 25), xytext=(5.5, 27),
             arrowprops=dict(facecolor="black", shrink=0.05))

plt.show()

# Reset style back to default for future plots in this notebook
plt.style.use("default")
```

## 🏆 Final Challenge — Put It All Together

**Scenario:** You're building a small "Class Performance Dashboard" for a teacher.

1. Create data for 6 students with scores in Math, Science, and English (make up realistic numbers)
2. Build a **2x2 subplot dashboard** containing:
   - Top-left: A grouped bar chart comparing all 3 subjects for all 6 students
   - Top-right: A pie chart showing the CLASS AVERAGE contribution of each subject to total marks
   - Bottom-left: A histogram of all combined scores (18 total values) showing the score distribution
   - Bottom-right: A scatter plot of Math score (x) vs Science score (y) for each student
3. Give the whole figure a main title using `fig.suptitle("Class Performance Dashboard")`
4. Use `plt.tight_layout()` so nothing overlaps
5. Save the final dashboard as `class_dashboard.png`

*(This combines bar charts, pie charts, histograms, scatter plots, subplots, and saving — everything from this notebook!)*

```python
# 👉 Student: write your Final Challenge solution here

import matplotlib.pyplot as plt
import numpy as np

# 1. Create data for 6 students
students = ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Usman', 'Fatima']

math = [85, 90, 78, 92, 75, 88]
science = [80, 85, 82, 88, 79, 91]
english = [78, 88, 80, 85, 82, 90]

# 2. Combine all scores
all_scores = math + science + english

# Create a 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# TOP-LEFT: Grouped bar chart
x = np.arange(len(students))
width = 0.25

axes[0, 0].bar(x - width, math, width, label='Math')
axes[0, 0].bar(x, science, width, label='Science')
axes[0, 0].bar(x + width, english, width, label='English')

axes[0, 0].set_title('Student Scores by Subject')
axes[0, 0].set_xlabel('Students')
axes[0, 0].set_ylabel('Scores')
axes[0, 0].set_xticks(x)
axes[0, 0].set_xticklabels(students)
axes[0, 0].legend()

# TOP-RIGHT: Pie chart of class average contribution
subject_averages = [
    np.mean(math),
    np.mean(science),
    np.mean(english)
]

axes[0, 1].pie(
    subject_averages,
    labels=['Math', 'Science', 'English'],
    autopct='%1.1f%%',
    startangle=90
)
axes[0, 1].set_title('Class Average Contribution by Subject')

# BOTTOM-LEFT: Histogram of all 18 scores
axes[1, 0].hist(all_scores, bins=8, edgecolor='black')
axes[1, 0].set_title('Distribution of All Student Scores')
axes[1, 0].set_xlabel('Score')
axes[1, 0].set_ylabel('Frequency')

# BOTTOM-RIGHT: Scatter plot of Math vs Science
axes[1, 1].scatter(math, science, color='red', alpha=0.7)

# Label each student's point
for i, student in enumerate(students):
    axes[1, 1].annotate(
        student,
        (math[i], science[i]),
        xytext=(5, 5),
        textcoords='offset points'
    )

axes[1, 1].set_title('Math vs Science Scores')
axes[1, 1].set_xlabel('Math Score')
axes[1, 1].set_ylabel('Science Score')

# 3. Add the main title
fig.suptitle('Class Performance Dashboard', fontsize=16)

# 4. Adjust layout
plt.tight_layout(rect=[0, 0, 1, 0.96])

# 5. Save the dashboard
plt.savefig('class_dashboard.png', dpi=200, bbox_inches='tight')

# Display the dashboard
plt.show()
```
